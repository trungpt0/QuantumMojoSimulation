```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Operator
import numpy as np
```

```python
qc1 = QuantumCircuit(4)
qc1.t(0)
qc1.cx(0,1)
qc1.cx(2,3)
qc1.t(3)
qc1.cx(2,3)
qc1.cx(1,2)
qc1.cx(1,0)
qc1.cx(3,2)
qc1.cx(1,2)
qc1.cx(0,1)
qc1.t(2)
qc1.tdg(1)

print(qc1.draw(output="text", fold=-1))
```

```python
qc3 = QuantumCircuit(4)
qc3.cx(0,1)
qc3.cx(2,3)
qc3.cx(2,3)
qc3.cx(1,2)
qc3.cx(1,0)
qc3.cx(3,2)
qc3.cx(1,2)
qc3.cx(0,1)
qc3.s(2)
print(qc3.draw(output="text", fold=-1))
```

```python
U1 = Operator(qc1).data
U2 = Operator(qc3).data

F = abs(np.trace(U1.conj().T @ U2))/16

print(F)
print(np.allclose(U1, U2))
```
